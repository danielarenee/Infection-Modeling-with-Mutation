```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from ipywidgets import interact, FloatSlider, IntSlider
import sys
from pathlib import Path

# Ensure sircmw_I_utils can be imported
notebook_dir = Path.cwd()
if notebook_dir.name == 'timeseries':
    sys.path.append(str(notebook_dir.parent))
else:
    sys.path.append(str(notebook_dir))

from sircmw_I_utils import sircmw, SIGMA

def plot_interactive_timeseries(beta=600, tilde_eps=0.5):
    # Setup parameters
    scale_factor = 0.001
    eps = tilde_eps / scale_factor
    p = {'beta0': float(beta), 'sigma': SIGMA, 'eps': eps}
    
    # Initial conditions
    y0 = np.array([0.2, 0.001, 0.499, 0.3])
    
    # Solve system using DOP853 over 100 years
    sol = solve_ivp(sircmw, (0.0, 100.0), y0, args=(p,),
                    method='DOP853', rtol=1e-6, atol=1e-9)
    
    # Plotting panels for each class
    fig, axes = plt.subplots(2, 2, figsize=(12, 8), dpi=120)
    titles = ['S (Susceptible)', 'I (Infected)', 'R (Recovered)', 'C (Cross-Immune)']
    colors = ['#4A90E2', '#D0021B', '#F5A623', '#7ED321']
    
    t_arr = sol.t
    for k in range(4):
        ax = axes[k // 2, k % 2]
        ax.plot(t_arr, sol.y[k], color=colors[k], linewidth=1.5)
        ax.set_title(titles[k], fontsize=12, fontweight='bold')
        ax.set_xlabel('Time (years)', fontsize=10)
        ax.set_ylabel('Proportion', fontsize=10)
        ax.grid(True, alpha=0.3)
        ax.set_xlim(0, 100)
        
    plt.tight_layout()
    plt.show()

# Setup interactive widgets
interact(
    plot_interactive_timeseries,
    beta=IntSlider(min=100, max=1000, step=50, value=600, description=r'$\beta_0$'),
    tilde_eps=FloatSlider(min=0.0, max=5.0, step=0.05, value=0.5, description=r'$\tilde{\varepsilon}$')
)
```
